# Model Performance Analysis

## Purpose

This notebook focuses on analyzing the performance of the CNN-based facial age group classification models developed in the ENGR 418 project. Instead of retraining the models, this analysis uses existing experimental results to better understand classification performance, class imbalance, and prediction errors.

## Analysis Performed

The following analyses were conducted:

1. **Class Distribution Analysis:** Examined age group distributions in the training, validation, and test sets across Face-Age-10K and UTKFace, using both 4-class and 8-class configurations.

2. **Training Samples vs. F1-score:** Compared the number of training samples in each age group with its corresponding test F1-score to investigate the relationship between sample availability and classification performance.

3. **Confusion Matrix Analysis:** Examined raw and normalized confusion matrices to identify which age groups were frequently misclassified.

4. **Prediction Bias Analysis:** Compared actual and predicted age group distributions to identify potential biases in model predictions.

## Key Findings

- The 8-class classification task generally achieved lower accuracy and Macro-F1 scores than the 4-class task.
- UTKFace showed substantial class imbalance, particularly in the 8-class configuration.
- The models struggled to distinguish several middle-aged and older age groups.
- The UTKFace 8-class model showed a strong tendency to overpredict the 31–40 age group while underpredicting several other groups.
- A larger number of training samples did not always correspond to better classification performance.

## Next Steps

The findings suggest that class imbalance and differences in age-related facial features may contribute to classification errors. Further investigation using class weighting is proposed to evaluate whether adjusting the contribution of different age groups during training can improve Macro-F1 and reduce prediction bias.

**Note:** The observed relationships do not establish causation. Additional controlled experiments are required to determine the effects of class weighting and other training strategies

```python
from pathlib import Path
from PIL import Image
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms

from sklearn.metrics import accuracy_score, f1_score

# Reproducibility
SEED = 418

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Project folder
PROJECT_FOLDER = Path.cwd()

# Dataset folder
UTKFACE_FOLDER = (
    PROJECT_FOLDER /
    "UTKFace_source" /
    "extracted" /
    "UTKFace"
)

# Existing split file
SPLIT_FILE = PROJECT_FOLDER / "data/splits/UTKFace_8Groups_split_indices.csv"

# Device
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Project folder:", PROJECT_FOLDER)
print("UTKFace folder exists:", UTKFACE_FOLDER.exists())
print("Split file exists:", SPLIT_FILE.exists())
print("Device:", device)
```

```python
# Load the existing train/validation/test split
split_table = pd.read_csv(SPLIT_FILE)

print("Total samples:", len(split_table))
print("\nColumns:")
print(split_table.columns.tolist())

print("\nSplit distribution:")
print(split_table["split"].value_counts())

print("\nFirst five rows:")
display(split_table.head())
```

```python
# Define four age groups

LABEL_NAMES = [
    "age 01-20",
    "age 21-40",
    "age 41-65",
    "age 66+"
]

def assign_age_group(age):
    if 1 <= age <= 20:
        return "age 01-20"
    elif 21 <= age <= 40:
        return "age 21-40"
    elif 41 <= age <= 65:
        return "age 41-65"
    else:
        return "age 66+"

# Create four-class labels
split_table["age_group_4"] = split_table["age"].apply(
    assign_age_group
)

print("Age group distribution:")
print(split_table["age_group_4"].value_counts())

print("\nTraining set distribution:")
print(
    split_table[split_table["split"] == "train"]
    ["age_group_4"]
    .value_counts()
    .reindex(LABEL_NAMES)
)

print("\nLabel names:", LABEL_NAMES)
```

```python
# Image preprocessing
IMAGE_SIZE = 200
BATCH_SIZE = 64

transform = transforms.Compose([
    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
    transforms.ToTensor()
])

# Convert age group names into numerical labels
label_to_id = {
    label: index
    for index, label in enumerate(LABEL_NAMES)
}

split_table["label_id"] = split_table["age_group_4"].map(
    label_to_id
)

# Custom Dataset
class UTKFaceDataset(Dataset):

    def __init__(self, dataframe, transform=None):
        self.dataframe = dataframe.reset_index(drop=True)
        self.transform = transform

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):
        row = self.dataframe.iloc[index]

        image = Image.open(row["path"]).convert("RGB")
        label = int(row["label_id"])

        if self.transform:
            image = self.transform(image)

        return image, label


# Separate datasets
train_table = split_table[
    split_table["split"] == "train"
].copy()

validation_table = split_table[
    split_table["split"] == "validation"
].copy()

test_table = split_table[
    split_table["split"] == "test"
].copy()

train_dataset = UTKFaceDataset(train_table, transform)
validation_dataset = UTKFaceDataset(validation_table, transform)
test_dataset = UTKFaceDataset(test_table, transform)

# DataLoaders
train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0
)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0
)

print("Training samples:", len(train_dataset))
print("Validation samples:", len(validation_dataset))
print("Test samples:", len(test_dataset))
print("Number of classes:", len(LABEL_NAMES))
```

```python
class SmallAgeCNN(nn.Module):

    def __init__(self, number_of_classes):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.AdaptiveAvgPool2d((1, 1))
        )

        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(p=0.30),
            nn.Linear(128, number_of_classes)
        )

    def forward(self, images):
        features = self.features(images)
        return self.classifier(features)


# Initialize model
model = SmallAgeCNN(
    number_of_classes=len(LABEL_NAMES)
).to(device)

print(model)

trainable_parameters = sum(
    p.numel() for p in model.parameters()
    if p.requires_grad
)

print("\nTrainable parameters:", trainable_parameters)
```

```python
def evaluate_model(model, data_loader):

    model.eval()

    all_predictions = []
    all_labels = []

    running_loss = 0.0
    samples_seen = 0

    with torch.no_grad():

        for images, labels in data_loader:

            images = images.to(device)
            labels = labels.to(device)

            logits = model(images)

            loss = nn.CrossEntropyLoss()(logits, labels)

            predictions = torch.argmax(logits, dim=1)

            running_loss += loss.item() * labels.size(0)
            samples_seen += labels.size(0)

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.cpu().numpy()
            )

    average_loss = running_loss / samples_seen

    accuracy = accuracy_score(
        all_labels,
        all_predictions
    )

    macro_f1 = f1_score(
        all_labels,
        all_predictions,
        average="macro",
        zero_division=0
    )

    return {
        "loss": average_loss,
        "accuracy": accuracy,
        "macro_f1": macro_f1
    }
```

```python
# ==========================================
# Experiment A: Standard CrossEntropyLoss
# ==========================================

# Reset random seeds
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Reinitialize model
model_A = SmallAgeCNN(
    number_of_classes=len(LABEL_NAMES)
).to(device)

# Standard CrossEntropyLoss (no class weights)
loss_function_A = nn.CrossEntropyLoss()

# Optimizer
optimizer_A = torch.optim.Adam(
    model_A.parameters(),
    lr=0.001
)

EPOCHS = 10

best_validation_macro_f1_A = -1.0

best_model_path_A = (
    PROJECT_FOLDER / "models/UTKFace_4Groups_Standard_CE_best.pt"
)

training_history_A = []

# Training loop
for epoch in range(1, EPOCHS + 1):

    model_A.train()

    running_loss = 0.0
    images_seen = 0

    for batch_images, batch_labels in train_loader:

        batch_images = batch_images.to(device)
        batch_labels = batch_labels.to(device)

        optimizer_A.zero_grad()

        logits = model_A(batch_images)

        loss = loss_function_A(
            logits,
            batch_labels
        )

        loss.backward()
        optimizer_A.step()

        running_loss += loss.item() * batch_labels.size(0)
        images_seen += batch_labels.size(0)

    train_loss = running_loss / images_seen

    validation_results = evaluate_model(
        model_A,
        validation_loader
    )

    training_history_A.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "validation_loss": validation_results["loss"],
        "validation_accuracy": validation_results["accuracy"],
        "validation_macro_f1": validation_results["macro_f1"]
    })

    # Save the best model based on validation Macro-F1
    if validation_results["macro_f1"] > best_validation_macro_f1_A:

        best_validation_macro_f1_A = (
            validation_results["macro_f1"]
        )

        torch.save({
            "model_state_dict": model_A.state_dict(),
            "label_names": LABEL_NAMES,
            "seed": SEED,
            "validation_macro_f1": best_validation_macro_f1_A,
            "image_size": [200, 200],
            "architecture": "SmallAgeCNN",
            "loss_function": "Standard CrossEntropyLoss"
        }, best_model_path_A)

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {validation_results['loss']:.4f} | "
        f"Val Accuracy: {validation_results['accuracy']:.4f} | "
        f"Val Macro-F1: {validation_results['macro_f1']:.4f}"
    )

# Save training history
history_table_A = pd.DataFrame(training_history_A)

history_csv_A = (
    PROJECT_FOLDER / "results/tables/UTKFace_4Groups_Standard_CE_history.csv"
)

history_table_A.to_csv(history_csv_A, index=False)

print("\nExperiment A completed!")
print(
    "Best validation Macro-F1:",
    round(best_validation_macro_f1_A, 4)
)

print("Best model:", best_model_path_A)
print("Training history:", history_csv_A)
```

```python
# Calculate class weights using training data only

train_label_ids = train_table["label_id"].to_numpy()

train_counts = np.bincount(
    train_label_ids,
    minlength=len(LABEL_NAMES)
)

class_weights = (
    len(train_label_ids) /
    (len(LABEL_NAMES) * train_counts)
)

class_weights_tensor = torch.tensor(
    class_weights,
    dtype=torch.float32,
    device=device
)

print("Class weights:")

for label, count, weight in zip(
    LABEL_NAMES,
    train_counts,
    class_weights
):
    print(
        f"{label}: "
        f"samples={count}, "
        f"weight={weight:.4f}"
    )
```

```python
# ==========================================
# Experiment B: Weighted CrossEntropyLoss
# ==========================================

# Reset random seeds
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

# Reinitialize model
model_B = SmallAgeCNN(
    number_of_classes=len(LABEL_NAMES)
).to(device)

# Weighted CrossEntropyLoss
loss_function_B = nn.CrossEntropyLoss(
    weight=class_weights_tensor
)

# Optimizer
optimizer_B = torch.optim.Adam(
    model_B.parameters(),
    lr=0.001
)

EPOCHS = 10

best_validation_macro_f1_B = -1.0

best_model_path_B = (
    PROJECT_FOLDER / "models/UTKFace_4Groups_Weighted_CE_best.pt"
)

training_history_B = []

# Training loop
for epoch in range(1, EPOCHS + 1):

    model_B.train()

    running_loss = 0.0
    images_seen = 0

    for batch_images, batch_labels in train_loader:

        batch_images = batch_images.to(device)
        batch_labels = batch_labels.to(device)

        optimizer_B.zero_grad()

        logits = model_B(batch_images)

        loss = loss_function_B(
            logits,
            batch_labels
        )

        loss.backward()
        optimizer_B.step()

        running_loss += loss.item() * batch_labels.size(0)
        images_seen += batch_labels.size(0)

    train_loss = running_loss / images_seen

    validation_results = evaluate_model(
        model_B,
        validation_loader
    )

    training_history_B.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "validation_loss": validation_results["loss"],
        "validation_accuracy": validation_results["accuracy"],
        "validation_macro_f1": validation_results["macro_f1"]
    })

    # Save best model based on validation Macro-F1
    if validation_results["macro_f1"] > best_validation_macro_f1_B:

        best_validation_macro_f1_B = (
            validation_results["macro_f1"]
        )

        torch.save({
            "model_state_dict": model_B.state_dict(),
            "label_names": LABEL_NAMES,
            "seed": SEED,
            "validation_macro_f1": best_validation_macro_f1_B,
            "image_size": [200, 200],
            "architecture": "SmallAgeCNN",
            "loss_function": "Weighted CrossEntropyLoss"
        }, best_model_path_B)

    print(
        f"Epoch {epoch:02d}/{EPOCHS} | "
        f"Train Loss: {train_loss:.4f} | "
        f"Val Loss: {validation_results['loss']:.4f} | "
        f"Val Accuracy: {validation_results['accuracy']:.4f} | "
        f"Val Macro-F1: {validation_results['macro_f1']:.4f}"
    )

# Save training history
history_table_B = pd.DataFrame(training_history_B)

history_csv_B = (
    PROJECT_FOLDER / "results/tables/UTKFace_4Groups_Weighted_CE_history.csv"
)

history_table_B.to_csv(history_csv_B, index=False)

print("\nExperiment B completed!")

print(
    "Best validation Macro-F1:",
    round(best_validation_macro_f1_B, 4)
)

print("Best model:", best_model_path_B)
print("Training history:", history_csv_B)
```

Overall, class weighting improved the balance of 
classification performance across age groups, although 
some performance was sacrificed in the majority class.
